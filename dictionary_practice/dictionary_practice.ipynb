```python
# Exercise 1: basic dictionary with key access
prices = {
    "water": 10000,
    "orange_juice": 20000,
    "redbull": 30000,
}

print(prices["orange_juice"])
print(prices["redbull"])
print(prices["water"])
```

```python
# Exercise 2: safe access with .get()
prices = {
    "water": 10000,
    "orange_juice": 20000,
    "redbull": 30000,
}

item = input("Which item do you want the price for ")
print(prices.get(item,"not found"))
```

```python
# Exercise 3: loop over dictionary with .items()
prices = {
    "water": 10000,
    "orange_juice": 20000,
    "redbull": 30000,
}

for item, price in prices.items():
    print(item, ":", price)
```
