```python
inventory = {}

while True:
    print("1. Add item")
    print("2. View inventory")
    print("3. Check price of an item")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        print("You chose Add")
        name = input("Item name: ")
        price = float(input("Item price: "))
        inventory[name] = price
        
    elif choice == "2":
        print("You chose View")
        for item, price in inventory.items():
            print(item, ":", price)

    elif choice == "3":
        print("You chose Check price")
        name = input("Item name: ")
        print(inventory.get(name, "Not found"))

    elif choice == "4":
        print("Goodbye")
        break


    
```
