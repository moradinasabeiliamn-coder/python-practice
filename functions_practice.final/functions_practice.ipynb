```python
# Exercise 1: function with no parameters
def greet(name, last_name, age):
    print("Welcome to my program!",name, last_name , age)

greet("Iliya", "moradi", "18")
greet("Sara", "Ahmadi", "22")
greet("Ali", "Hosseini", "30")

# Exercise 2: function with parameters and print
def numbers(a, b):
    print("result =", a + b)

numbers(5, 8)


# Exercise 3: function with return
def numbers_return(a, b):
    return a + b

result = numbers_return(5, 8)
print("result =", result)


# Exercise 4: function that takes a list and returns a sum
def total_price(prices):
    return sum(prices)

my_prices = [10, 20, 30]
result = total_price(my_prices)
print("Total:", result)


# Exercise 5: function with return + if/else
def check_grade(grade):
    if grade >= 10:
        return "pass"
    else:
        return "fail"

result = check_grade(15)
print(result)

result2 = check_grade(9)
print(result2)
```
