```python
shopping_list = []

while True:
    print("1. Add item")
    print("2. Remove item")
    print("3. View list")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        shopping_list.append(input("What item do you want to add: "))
        print(shopping_list)

    elif choice == "2":
        shopping_list.remove(input("What item do you want to remove: "))
        print(shopping_list)

    elif choice == "3":
        count = 0
        for item in shopping_list :
            count = 1 + count
            print(count, "." , item)
        
    elif choice == "4":
        print("Goodbye")
        break
```
